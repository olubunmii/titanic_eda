```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
titanic = pd.read_csv('/Users/olubunmi/Desktop/Titanic_EDA/Data/Titanic.csv')
titanic.head(5)
#titanic.shape
```

```python
titanic['Pclass'].value_counts()
```

```python
#titanic[titanic['Pclass']== 1]
class_1df = titanic[titanic['Pclass'] == 1]
totalClass1 = class_1df['Survived'].value_counts().sum()
totalClass1
```

```python
survived1 = class_1df['Survived'] == 1
survivedClass1 = survived1.sum()
survivedClass1
```

```python
survivalRateInClass1 = ((survivedClass1/totalClass1)*100).round(2)
survivalRateInClass1
```

```python
class_2df = titanic[titanic['Pclass'] == 2]
totalClass2 = class_2df['Survived'].value_counts().sum()
totalClass2
```

```python
survived2 = class_2df['Survived'] == 1
survivedClass2 = survived2.sum()
survivedClass2
```

```python
survivalRateInClass2 = ((survivedClass2/totalClass2)*100).round(2)
survivalRateInClass2
```

```python
class_3df = titanic[titanic['Pclass'] == 3]
totalClass3 = class_3df['Survived'].value_counts().sum()
totalClass3
```

```python
survived3 = class_3df['Survived'] == 1
survivedClass3 = survived3.sum()
survivedClass3
```

```python
survivalRateInClass3 = ((survivedClass3/totalClass3)*100).round(2)
survivalRateInClass3
```

```python
survivalRateInClass1, survivalRateInClass2, survivalRateInClass3
```

```python
df = pd.DataFrame({
    'class':['class1','class2','class3'],
    'survival rate':[survivalRateInClass1,survivalRateInClass2,survivalRateInClass3]
})

df
```

```python
sns.barplot(data = df, x = "class", y = "survival rate")
plt.savefig("SurvivalvsClass.png")
```

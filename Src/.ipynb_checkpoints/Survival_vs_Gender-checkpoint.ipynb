# **HYPOTHESIS 2: Determine if the survival rate is associated to the gender of passenger**

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
titanic = pd.read_csv('/Users/olubunmi/Desktop/Titanic_EDA/Data/Titanic.csv')
titanic.head(5)
#titanic.shape
```

```python
titanic[['Sex', 'Survived']].value_counts()
```

```python
total_men = titanic[titanic['Sex'] == 'male']['Survived'].value_counts().sum()
total_men
```

```python
men_survived = titanic[(titanic['Sex'] == 'male') & (titanic['Survived'] == 1)][['Sex', 'Survived']].value_counts().sum()
men_survived
```

```python
survRateMen = ((men_survived/total_men)*100).round(2)
survRateMen
```

```python
total_women = titanic[titanic['Sex'] == 'female']['Survived'].value_counts().sum()
total_women
```

```python
women_survived = titanic[(titanic['Sex'] == 'female') & (titanic['Survived'] == 1)][['Sex', 'Survived']].value_counts().sum()
women_survived
```

```python
survRateWomen = ((women_survived/total_women)*100).round(2)
survRateWomen
```

```python
survRateMen, survRateWomen
```

```python
surv_genderdf = pd.DataFrame({
    'gender':['male','female'],
    'gender survival rate':[survRateMen,survRateWomen]
})

surv_genderdf
```

```python
sns.barplot(data=surv_genderdf, x='gender', y='gender survival rate', hue='gender survival rate', palette='Set2')
plt.title('Survival Count by Gender')
plt.savefig("SurvivalvsGender.png")
plt.show()
```

# **HYPOTHESIS 3: Determine if the survival rate is associated to the age of passenger**

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
titanic = pd.read_csv('/Users/olubunmi/Desktop/Titanic_EDA/Data/Titanic.csv')
titanic.head(5)
#titanic.shape
```

```python
titanic.sort_values('Age')
```

```python
children_df = titanic[titanic['Age']<= 19]
total_children = children_df['Age'].value_counts().sum()
#children_df
total_children
```

```python
surv_children = (children_df['Survived']==1).sum()
surv_children
```

```python
children_survR = ((surv_children/total_children)*100).round(2)
children_survR
```

```python
adults_df = titanic[(titanic['Age']>= 20) & (titanic['Age'] <=59)]
total_adults= adults_df['Age'].value_counts().sum()
#adults_df['Age'].sort_values()
total_adults
```

```python
surv_adults = (adults_df['Survived'] == 1).sum()
surv_adults
```

```python
adults_survR = ((surv_adults/total_adults)*100).round(2)
adults_survR
```

```python
seniors_df = titanic[titanic['Age'] >= 60]
total_seniors = seniors_df['Age'].value_counts().sum()
total_seniors
```

```python
surv_seniors = (seniors_df['Survived']==1).sum()
surv_seniors
```

```python
seniors_survR = ((surv_seniors/total_seniors)*100).round(2)
seniors_survR
```

```python
children_survR, adults_survR, seniors_survR
```

```python
x = np.array(["Children", "Adults", "Seniors"])
y = np.array([children_survR, adults_survR, seniors_survR])
#plt.barh(x, y, color = 'green')
plt.title("Survival Rate according to Age")
plt.xlabel("Survival Rate (%)")

colors = ['purple', 'orange', 'pink']
bars = plt.barh(x, y, color=colors)

# Display percentages on each bar
plt.bar_label(bars, fmt='%.1f%%', padding=5)
plt.savefig("SurvivalvsAge.png")
plt.show()
```
